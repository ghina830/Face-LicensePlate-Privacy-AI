```python
class plates_dataset(datasets.ImageFolder):
  def __init__(self,img_dir,transform=None):
    super().__init__(img_dir,transform=transform)
    self.class_idx={'license_plates': 0.0, 'non-license_plates': 1.0}
  def __getitem__(self, idx):
    img_path,_=self.samples[idx]
    class_name=os.path.basename(os.path.dirname(img_path))

    img_label=self.class_idx[class_name]
    image= self.loader(img_path)
    if self.transform is not None:
      image= self.transform(image)
    return image,img_label
```

```python
import torch.nn.functional as F
class plate_modified_VGG16(nn.Module):
  def __init__(self):
    super().__init__()
    self.backbone=models.vgg16(pretrained=True).features
    #self.avgpool = nn.AdaptiveAvgPool2d(output_size=(7, 7))
    self.new_fc_layer=nn.Sequential(
      nn.Linear(512, 32),
      nn.ReLU(inplace=True),
      nn.Dropout(p=0.2,inplace=False),
      nn.Linear(32, 64),
      nn.ReLU(inplace=True),
      nn.Dropout(p=0.2,inplace=False),
      nn.Linear(64, 1),
      nn.Sigmoid()
    )
    for param in self.backbone.parameters():
       param.requires_gards = False

  def forward(self, x):
        x = self.backbone(x)
        x = F.max_pool2d(x,kernel_size=x.size()[2:])
        x = x.view(x.size(0), -1)
        x = F.dropout(x, 0.5)
        x = self.new_fc_layer(x)
        x=x.squeeze(1)

        return x
```

```python
plate_modified_VGG16()
```

```python
def train(dataloader, model, loss_fn, optimizer,current_epoch):
    size = len(dataloader.dataset)
    correct = 0

    model.train()
    for batch, (X,label) in enumerate(dataloader):

        device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
        X = X.to(device)
        label = label.to(device).float()

        Y = model(X).float()
        loss = loss_fn(Y,label)
        loss.backward()
        optimizer.step()
        optimizer.zero_grad()


        predicted = torch.round(Y)
        correct += (predicted == label).sum().item()

        if batch % 100 == 0:
            loss, current = loss.item(), (batch + 1) * len(X)
            print(f"loss: {loss:>7f}  [{current:>5d}/{size:>5d}]")


    train_acc = 100 * correct / size
    print(f"Training Accuracy: {train_acc:.2f}%")

    path = f"drive/MyDrive/model_plates/model_epoch_{current_epoch}.pt"
    torch.save(model.state_dict(), path)
```

```python
def test(dataloader, model, loss_fn):

    model.eval()
    size = len(dataloader.dataset)
    num_batches = len(dataloader)
    test_loss = 0
    correct = 0
    with torch.no_grad():
        for X,label in dataloader:

            device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
            X = X.to(device)
            label = label.to(device).float()

            Y = model(X).float()
            test_loss += loss_fn(Y,label).item()
            correct += (torch.round(Y) == label).type(torch.float).sum().item()

    test_loss /= num_batches
    correct /= size

    print(f"Test Error: \n Accuracy: {(100*correct):>0.1f}%, Avg loss: {test_loss:>8f} \n")
```

```python
transform = transforms.Compose([
    transforms.Resize((256, 256)),
    transforms.ColorJitter(brightness=0.5, contrast=0.5, saturation=0.5, hue=0.5),
    transforms.RandomAffine(degrees=90, translate=(0.4, 0.4), scale=(0.6, 1.4), shear=15),
    transforms.ToTensor(),

])
```

```python
plates_data = plates_dataset('/content/pDataset',transform)

train_size = int(0.9 * len( plates_data))
test_size = len( plates_data ) - train_size
training_plates_data, testing_plates_data = data.random_split( plates_data , [train_size, test_size])


train_dataloader_plates=data.DataLoader(training_plates_data,batch_size=64,shuffle=True)
test_dataloader_plates=data.DataLoader(testing_plates_data,batch_size=64,shuffle=True)
```

```python
plates_model = plate_modified_VGG16()

device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
plates_model = plates_model.to(device)

optimizer = optim.Adam(plates_model.parameters(), lr=1e-4)
loss_fn = nn.BCELoss()

if not os.path.exists('drive/MyDrive/model_plates'):
    os.makedirs('drive/MyDrive/model_plates')

epochs = 20
for t in range(epochs):
    print(f"Epoch {t+1}\n-------------------------------")
    train(train_dataloader_plates, plates_model, loss_fn, optimizer,t+1)
    test(test_dataloader_plates, plates_model, loss_fn)


print("Done!")
```
